```python
import os
import minsearch
import pandas as pd

from dotenv import load_dotenv
```

```python
load_dotenv()
GEMINI_API_KEY =os.getenv("GEMINI_API_KEY")
```

```python
df =pd.read_csv("../data/processed/recipes_documents.csv")
index = minsearch.Index(
    text_fields=["document"],
    keyword_fields=["name"]
)

index.fit(df.to_dict(orient="records"))
```

```python
query = "I want a qucik vegitrain dinner"

result = index.search(
    query=query,
    num_results=5
)
result
```

```python
prompt_template = """
You are an AI Kitchen Assistant.

Answer the user's question using ONLY the recipe information below.

User Question:
{question}

Recipes:
{context}
""".strip()
```

```python
def build_prompt(query, search_results):
    context = ""

    for recipe in search_results:
        context = context + recipe["document"] + "\n\n"

    prompt = prompt_template.format(
        question=query,
        context=context
    )

    return prompt
```

```python
prompt = build_prompt(query, result)

print(prompt)
```

```python
from dotenv import load_dotenv
import os
from google import genai

load_dotenv()

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)
```

```python
def llm(prompt):
    response = client.models.generate_content(
         model="gemini-3.1-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
answer = llm(prompt)

print(answer)
```

```python
def rag(query):
    search_results = index.search(
        query=query,
        num_results=5
    )

    prompt = build_prompt(query, search_results)

    answer = llm(prompt)

    return answer
```

```python
print(rag("I want a healthy breakfast"))
```
