```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/RAW_recipes.csv")
```

```python
print("Dataset Shape:",df.shape)
```

```python
df.columns.tolist()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df= df.dropna(subset=["name"])
df["description"] = df["description"].fillna("")
```

```python
print("Duplicate rows",df.duplicated().sum())
```

```python
print(df.shape)
```

```python
columns=[
     "name",
     "description",
     "ingredients",
     "steps",
     "tags",
     "minutes"

]
df=df[columns]
```

```python
df.dtypes
```

```python
import ast
```

```python
df["steps"] = df["steps"].apply(ast.literal_eval)
df["tags"] = df["tags"].apply(ast.literal_eval)
```

```python
df.to_csv("../data/processed/recipes_clean.csv",index=False)
```

```python
df.to_pickle("../data/processed/recipes_clean.pkl")
```

```python
def create_document(row):
    return f"""
Recipe Name: {row['name']}

Description:
{row['description']}

Ingredients:
{', '.join(row['ingredients'])}

Steps:
{' '.join(row['steps'])}

Tags:
{', '.join(row['tags'])}

Cooking Time:
{row['minutes']} minutes
"""
```

```python
df["document"] = df.apply(create_document, axis=1)
```

```python
print(df["document"][0])
```

```python
df.to_csv("../data/processed/recipes_documents.csv",index=False)
```
